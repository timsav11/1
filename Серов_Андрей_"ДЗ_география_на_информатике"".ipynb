## Задание 1. Расстояние между двумя точками на глобусе (2 балла)

Даны две точки на поверхности Земли, заданные своими координатами:
- широта (`lat`) и долгота (`lon`) в градусах.

Напишите функцию `distance_on_the_globe(lat1, lon1, lat2, lon2)`, которая возвращает расстояние в километрах.
Проверьте себя, посчитав расстояние между Санкт-Петербургом и Москвой.

```python
def distance_on_the_globe(lat1, lon1, lat2, lon2):
    """
    Вычисляет расстояние между двумя точками на сфере (в км).
    Координаты передаются в градусах.
    """
    lat1, lon1, lat2, lon2 = map(math.radians, [lat1, lon1, lat2, lon2])
    a = math.sin((lat2 - lat1) / 2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin((lon2 - lon1) / 2)**2
    dist = 6371 * 2 * math.asin(math.sqrt(a))

    return dist


# Расстояние между Москвой (55.7558, 37.6173) и Санкт-Петербургом (59.9343, 30.3351) примерно 635 км.
print(round(distance_on_the_globe(55.7558, 37.6173, 59.9343, 30.3351), 2), "км")
```

**Вопрос:** как зависит расстояние между точками от широты?

1. Зафиксировать разность долгот `Δlon = 10°`.
2. Создать массив широт от −90° до +90° с шагом 1°.
3. Для каждой широты `φ` посчитать расстояние между точками `(φ, 0°)` и `(φ, 10°)` через вашу функцию `distance_on_the_globe`.

```python


# Фиксируем разность долгот
delta_lon = 10

lats = range(-90,90)

# Считаем расстояния
distances = []
for lat in lats:
    d = distance_on_the_globe(lat,0,lat,delta_lon)
    distances.append(d)


# TODO: построение графика
plt.plot(lats, distances)
plt.title("Зависимость расстояния от широты (Δlon = 10°)")
plt.xlabel("Широта, градусы")
plt.ylabel("Расстояние, км")
plt.grid()
plt.show()
```
